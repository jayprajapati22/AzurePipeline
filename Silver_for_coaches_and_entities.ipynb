
Dynamic Data Reading



```python
#Parameters

dbutils.widgets.text("source_folder", "")
dbutils.widgets.text("sink_container", "")
dbutils.widgets.text("folder", "")
```

```python
source_container = dbutils.widgets.get("source_container")
sink_container = dbutils.widgets.get("sink_folder")
folder = dbutils.widgets.get("folder")
```


Parameterizing Code

```python
df = spark.read.format("parquet")\
                .load(f"abfss://{source_container}@dbpocstorage.dfs.core.windows.net/{folder}")
```

```python
df.display()
```

```python
df.write.format("delta")\
            .mode("append")
            .option("path", f"abfss://{sink_container}@azuredatastor.dfs.core.windows.net/{folder}")
            .saveAsTable(f"olympics.{sink_container}.{folder}")
```
